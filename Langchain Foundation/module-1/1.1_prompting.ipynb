```
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from pprint import pprint
from rich.console import Console
from rich.markdown import Markdown
from rich.panel import Panel
from pydantic import BaseModel
```

```
model = ChatOllama(model='gemma4:e4b', temperature=0)
agent = create_agent(
    model=model,
)
```

```
question = HumanMessage(content='Whats the capital of moon?')
response = agent.invoke({"messages":[question]})
pprint(response['messages'][-1].content)
```

```
system_prompt = 'You are a science fiction writer, create a capital city at the users request'
scifi_agent = create_agent(
    model=model,
    system_prompt=system_prompt
)
response = scifi_agent.invoke({"messages":[question]})
pprint(response['messages'][-1].content)
```

```
system_prompt = """You are a science fiction writer, create a capital city at the users request
User: What is the capital of mars?
Scifi Writer: Marsialis

User: What is the capital of jupiter?
Scifi Writer: Jovialis

User: What is the capital of saturn?
Scifi Writer: Saturnalis

"""
question = HumanMessage(content='What is the capital of venus?')
scifi_agent = create_agent(
    model=model,
    system_prompt=system_prompt
)
response = scifi_agent.invoke({"messages":[question]})
pprint(response['messages'][-1].content)
```

```
system_prompt ="""
You are a science fiction writer, create a capital city at the users request
Please keep to the below structure.
Name: The name of the capital city
Location: Where it is based
Vibe: 2-3 words to describe the vibe of the city
Economy: Main industries

"""
console = Console()
scifi_agent = create_agent(
    model=model,
    system_prompt=system_prompt
)
response = scifi_agent.invoke({"messages":[question]})
md = Markdown(response['messages'][-1].content)
console.print(response['messages'][-1].content)
```

```
class ScifiCity(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str


agent = create_agent(
    model=model,
    system_prompt='You are a Science fiction writer, create a capital city at the users request',
    response_format=ScifiCity
)
question = HumanMessage(content='What is the capital of mars?')
response = agent.invoke({"messages":[question]})
content = response['messages'][-1].content
console.print(Panel(content, title='AI Response'))
```
