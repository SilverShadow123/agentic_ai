```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from langchain_core.tools import tool
```

```python
model = ChatOllama(
    model='gemma4:e4b',
)
agent = create_agent(
    model=model,
    system_prompt='You are a full-stack comedian',
)
```

```python
human_msg = HumanMessage('Hello, How are you?')

result = agent.invoke({'messages':[human_msg]})
print(result['messages'][-1].content)
```

```python
print(type(result['messages'][-1].content))
```

```python
for msg in result['messages']:
    print(f'{msg.type}: {msg.content}\n')
```

```python
messages = [
    {"role": "system", "content": "You are a sports poetry expert who completes haikus that have been started"},
    {"role": "user", "content": "Write a haiku about sprinters"},
    {"role": "assistant", "content": "Feet don't fail me..."}
]
result = agent.invoke({'messages':messages})
print(result['messages'][-1].content)
```

```python
@tool
def check_haiku_lines(text: str):
    """Check if the given haiku text has exactly 3 lines.
    Returns None if its's correct, otherwise an error message.
    """
    lines = [line.strip() for line in text.strip().splitlines() if line.strip()]
    print(f'checking haiku it has {len(lines)} lines:\n {text}')

    if len(lines) != 3:
        return f'Incorrect! Haiku must have 3 lines, but has {len(lines)} lines'
    return 'Correct, haiku has 3 lines'
```

```python
agent = create_agent(
    model=model,
    tools=[check_haiku_lines],
    system_prompt='You are a sports poet who only writes Haiku. You always check your work.',
)
```

```python
result = agent.invoke({'messages':'Please write me a poet'})
```
